Q5. Reverse Every Alternate Digit

```python
num = 12345678
temp = num
total_digits = 0
while temp > 0:
    total_digits += 1
    temp //= 10

extracted = []
temp = num
for i in range(total_digits, 0, -1):
    digit = temp % 10
    temp //= 10
    if i % 2 == 0:
        extracted.append(digit)

result = 0
temp = num
extract_idx = 0
multiplier = 1
for i in range(total_digits, 0, -1):
    digit = temp % 10
    temp //= 10
    if i % 2 == 0:
        result += extracted[extract_idx] * multiplier
        extract_idx += 1
    else:
        result += digit * multiplier
    multiplier *= 10
print(result)
```

Q49. Check Fascinating Number

```python
num = 192
n2 = num * 2
n3 = num * 3

# Mathematically count digits to properly concatenate without str()
temp2, d2 = n2, 0
while temp2 > 0:
    d2 += 1
    temp2 //= 10

temp3, d3 = n3, 0
while temp3 > 0:
    d3 += 1
    temp3 //= 10

# Concatenate mathematically: num -> num*10^d2 + n2 -> ...
concat_num = (num * (10 ** d2) + n2) * (10 ** d3) + n3

# Check if exactly 1 through 9 are present using a mathematical positional tracker
temp = concat_num
digit_tracker = 0
while temp > 0:
    digit = temp % 10
    digit_tracker += 10 ** digit  # Increases the positional value for that digit
    temp //= 10

# 1111111110 means digits 1-9 appeared exactly once (10^1 + 10^2 ... + 10^9) and 0 didn't appear.
if digit_tracker == 1111111110:
    print("Fascinating")
else:
    print("Not Fascinating")
```

Q52. Check Smith Number

```python
num = 22
temp = num
sum_n = 0
# Sum of digits of the number
while temp > 0:
    sum_n += temp % 10
    temp //= 10

sum_factors = 0
temp = num
i = 2
is_composite = False

# Find prime factors and sum their digits
while temp > 1:
    if temp % i == 0:
        if num != i:  # Must be composite
            is_composite = True

        # Add digits of the prime factor
        factor = i
        while factor > 0:
            sum_factors += factor % 10
            factor //= 10

        temp //= i
    else:
        i += 1

if is_composite and sum_n == sum_factors:
    print("Smith")
else:
    print("Not Smith")
```

Q53. Check Keith Number

```python
num = 197
temp = num
total_digits = 0
while temp > 0:
    total_digits += 1
    temp //= 10

# Mathematically extract digits left to right into a sequence
sequence = []
temp = num
divisor = 10 ** (total_digits - 1)
while divisor > 0:
    sequence.append(temp // divisor)
    temp %= divisor
    divisor //= 10

# Generate next terms in the sequence
is_keith = False
while True:
    seq_sum = sum(sequence)
    if seq_sum == num:
        is_keith = True
        break
    if seq_sum > num:
        break

    # Shift sequence: remove first, append new sum
    sequence.pop(0)
    sequence.append(seq_sum)

if is_keith:
    print("Keith")
else:
    print("Not Keith")
```

Q54. Check Emirp Number

```python
num = 13

# Check if original is prime
is_prime = num > 1
for i in range(2, int(num ** 0.5) + 1):
    if num % i == 0:
        is_prime = False
        break

# Reverse the number mathematically
rev = 0
temp = num
while temp > 0:
    rev = (rev * 10) + (temp % 10)
    temp //= 10

# Check if reversed is prime
is_rev_prime = rev > 1
for i in range(2, int(rev ** 0.5) + 1):
    if rev % i == 0:
        is_rev_prime = False
        break

if is_prime and is_rev_prime and num != rev:
    print("Emirp")
else:
    print("Not Emirp")
```

Q55. Check Circular Prime

```python
num = 197
temp = num
digits = 0
while temp > 0:
    digits += 1
    temp //= 10

multiplier = 10 ** (digits - 1)
temp = num
is_circular = True

for _ in range(digits):
    # Check if current rotation is prime
    if temp <= 1:
        is_circular = False
        break
    for i in range(2, int(temp ** 0.5) + 1):
        if temp % i == 0:
            is_circular = False
            break

    if not is_circular:
        break

    # Rotate digits: move the last digit to the front
    last_digit = temp % 10
    temp = (last_digit * multiplier) + (temp // 10)

if is_circular:
    print("Circular Prime")
else:
    print("Not Circular Prime")
```

Q100. Catalan Numbers

```python
n = 5
for i in range(n):
    # Calculate factorial for 2i, i+1, and i
    fact_2i = 1
    for j in range(1, 2 * i + 1): fact_2i *= j

    fact_i_plus_1 = 1
    for j in range(1, i + 2): fact_i_plus_1 *= j

    fact_i = 1
    for j in range(1, i + 1): fact_i *= j

    catalan = fact_2i // (fact_i_plus_1 * fact_i)
    print(catalan, end=" ")
```

Q118. Add Two Binary Numbers

```python
bin1 = 1011
bin2 = 1101

def bin_to_dec(n):
    dec, base = 0, 1
    while n > 0:
        dec += (n % 10) * base
        base *= 2
        n //= 10
    return dec

def dec_to_bin(n):
    b, mult = 0, 1
    if n == 0: return 0
    while n > 0:
        b += (n % 2) * mult
        mult *= 10
        n //= 2
    return b

print(dec_to_bin(bin_to_dec(bin1) + bin_to_dec(bin2)))
```

Q125. Check Binary Palindrome

```python
num = 5
temp = num

# Build binary representation as integer
bin_val = 0
mult = 1
while temp > 0:
    bin_val += (temp % 2) * mult
    mult *= 10
    temp //= 2

# Reverse the binary integer
temp_bin = bin_val
rev_bin = 0
while temp_bin > 0:
    rev_bin = (rev_bin * 10) + (temp_bin % 10)
    temp_bin //= 10

if bin_val == rev_bin:
    print("Palindrome")
else:
    print("Not Palindrome")
```

Q144. Find Two Unique Numbers

```python
arr = [2, 4, 7, 9, 2, 4]
xor_all = 0
for n in arr:
    xor_all ^= n

# Get the rightmost set bit
rightmost_set_bit = xor_all & -xor_all

group1 = 0
group2 = 0
for n in arr:
    if n & rightmost_set_bit:
        group1 ^= n
    else:
        group2 ^= n

if group1 < group2:
    print(group1, group2)
else:
    print(group2, group1)
```

Q145. Reverse Bits of a Number

```python
num = 13
temp = num
bit_len = 0
# Find bit length
while temp > 0:
    bit_len += 1
    temp >>= 1

rev = 0
for i in range(bit_len):
    if num & (1 << i):
        rev |= (1 << ((bit_len - 1) - i))
print(rev)
```

Q146. Check Binary Palindrome Using Bit Operations

```python
num = 5
temp = num
bit_len = 0
while temp > 0:
    bit_len += 1
    temp >>= 1

rev = 0
for i in range(bit_len):
    if num & (1 << i):
        rev |= (1 << ((bit_len - 1) - i))

if num == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```
