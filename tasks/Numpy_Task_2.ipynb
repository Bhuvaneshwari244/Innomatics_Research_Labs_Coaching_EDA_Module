##VI. List vs Array

28. Explain the differences between Python lists and NumPy arrays in terms of memory, operations, and performance.

Memory: Python lists store pointers referencing scattered objects across memory; NumPy arrays store homogeneous data in single contiguous memory buffers.

Operations: Python lists concatenate on + and repeat on *; NumPy arrays perform vectorized, element-wise math directly.

Performance: NumPy operations run in optimized C routines with SIMD vectorization, executing significantly faster than Python for loops.

32. Explain why NumPy arrays are stored in contiguous blocks of memory and why this is important.

Explanation: Contiguous allocation stores elements in continuous adjacent memory slots. This allows CPU pre-fetchers to leverage CPU cache locality (reducing cache misses) and enables SIMD hardware vectorization to process multiple elements in a single clock cycle.

50. Why is 64-bit floating-point preferred for calculations over 16-bit floating-point?


Precision & Dynamic Range: float64 offers ~15–17 decimal digits of precision with an 11-bit exponent, whereas float16 offers only ~3–4 decimal digits with a 5-bit exponent.

Numerical Stability: In iterative or large-scale mathematical computations, float16 quickly leads to underflow, overflow, and accumulated rounding errors.

Hardware Support: Standard CPUs compute 64-bit floating-point math natively in registers with minimal overhead.

52. Explain the role of specifying the data type when creating an array.

Memory Optimization: Choosing smaller data types (e.g., np.uint8 for images or np.int16 vs np.int64) significantly reduces RAM usage.  

Preventing Data Loss & Overflow: Explicit typing prevents accidental type promotion and ensures hardware compatibility (e.g., matching GPU tensor expectations).

Performance: Smaller bitwidths maximize cache locality and increase vectorization efficiency (SIMD).

##X. Exploring Documentation

53. Explore the NumPy documentation and learn about all the array methods and functions that we have studied in class so far. Write a short note on what each method does and how it can be used.

**Array Creation Functions**

* **`np.array(object, dtype=None)`**: Converts Python sequences (like lists or tuples) into an N-dimensional NumPy array (`ndarray`) with an optional specified data type.


* **`np.zeros(shape, dtype=float)`**: Creates and returns a new array of the given shape, with every element initialized to zero (`0`).


* **`np.ones(shape, dtype=float)`**: Returns a new array of the specified shape and type, filled entirely with ones (`1`).


* **`np.empty(shape, dtype=float)`**: Allocates an array of the given shape without initializing its entries, which returns whatever uninitialized values happen to exist in that memory block.


* **`np.full(shape, fill_value, dtype=None)`**: Generates an array of a specified shape where every element is filled with a constant scalar value (`fill_value`).


* **`np.eye(N, M=None, k=0)`**: Creates a 2D matrix with ones along the $k$-th diagonal and zeros elsewhere.


* **`np.identity(n)`**: Produces an $n \times n$ square identity matrix with ones on the main diagonal and zeros elsewhere.


* **`np.zeros_like(a)` / `np.ones_like(a)` / `np.empty_like(a)**`: Creates a new array of zeros, ones, or uninitialized values that mirrors the shape and data type of the input array `a`.


* **`np.tril(m, k=0)` / `np.triu(m, k=0)**`: Returns the lower or upper triangular part of an array, zeroing out elements above or below the $k$-th diagonal.



**Sequencing & Range Functions**

* **`np.arange([start,] stop[, step,], dtype=None)`**: Generates evenly spaced numerical values within a half-open interval $[start, stop)$ using a defined step size.


* **`np.linspace(start, stop, num=50, endpoint=True)`**: Returns a specified count (`num`) of evenly spaced numbers over a closed or half-open interval $[start, stop]$.



**Array Inspection Attributes**

* **`.ndim`**: Displays the number of dimensions/axes of an array.


* **`.shape`**: Returns a tuple of integers showing the size of the array along each dimension.


* **`.size`**: Returns the total number of elements contained in the array.


* **`.dtype`**: Specifies the data type of the array elements (e.g., `int8`, `int64`, `float64`).


* **`.itemsize`**: Returns the memory footprint of a single element in bytes.


* **`.nbytes`**: Returns the total number of bytes consumed by the array data ($\text{size} \times \text{itemsize}$).


* **`.base`**: Identifies whether the array owns its own memory buffer or is a view referencing another base array.


* **`.ctypes.data`**: Provides the exact physical memory address pointer for the array's contiguous data buffer.



**Array Reshaping & Manipulation Methods**

* **`.reshape(shape, order='C')`**: Reconfigures the dimensions of an array without altering its underlying data, returning a view when possible.


* **`np.resize(a, new_shape)`**: Returns a new array with the specified shape, repeating copies of `a` if the new shape is larger.


* **`.astype(dtype)`**: Copies and casts the elements of an array to a newly specified data type.


* **`np.clip(a, a_min, a_max)`**: Restricts the values of an array to remain within the interval $[a\_min, a\_max]$.


* **`np.insert(arr, obj, values, axis=None)`**: Inserts values along a chosen axis before the given indices, returning a new copy.


* **`np.delete(arr, obj, axis=None)`**: Returns a new array with sub-arrays along the specified axis removed.


* **`.copy()`**: Creates a completely independent deep copy of the array in a newly allocated memory block.


* **`.view()`**: Creates a new array object that shares and references the exact same memory buffer as the original array.



**Random Number Generation Functions**

* **`np.random.randint(low, high=None, size=None)`**: Generates random whole numbers from the discrete uniform distribution over the range $[low, high)$.


* **`np.random.rand(d0, d1, ...)` / `np.random.random(size)**`: Generates random floating-point values from a uniform distribution over $[0.0, 1.0)$.


* **`np.random.uniform(low=0.0, high=1.0, size=None)`**: Samples random numbers from a defined continuous uniform distribution $[low, high)$.


* **`np.random.randn(d0, d1, ...)`**: Generates random floating-point numbers sampled from the standard normal (Gaussian) distribution ($\mu=0, \sigma=1$).


* **`np.random.normal(loc=0.0, scale=1.0, size=None)`**: Samples numbers from a normal distribution with a customized mean (`loc`) and standard deviation (`scale`).


* **`np.random.seed(seed)`**: Sets the random seed generator to ensure reproducible random outputs across runs.



**Statistical & Aggregate Functions**

* **`np.min(a)` / `.min()` & `np.max(a)` / `.max()**`: Calculates the minimum or maximum value across an array or along a designated axis.


* **`np.argmin(a)` / `np.argmax(a)**`: Returns the index position of the minimum or maximum element.


* **`np.mean(a)` / `.mean()**`: Computes the arithmetic mean of an array or along a specified axis.


* **`np.median(a)`**: Computes the median value along a specified axis.


* **`np.std(a)` / `.std()` & `np.var(a)` / `.var()**`: Calculates the standard deviation and variance across the array or along a given axis.


* **`np.percentile(a, q)` / `np.quantile(a, q)**`: Computes the $q$-th percentile or quantile of the data along the specified axis.


* **`np.nanmean(a)`, `np.nanmin(a)`, `np.nanmax(a)`, `np.nanstd(a)`, `np.nanvar(a)**`: Computes respective statistics while safely ignoring any `NaN` (Not a Number) values present in the array.



**Diagnostic Tool**

* **`np.info(object)`**: Outputs detailed internal diagnostic information about an array, including its class, shape, strides, data type, and memory contiguity.
