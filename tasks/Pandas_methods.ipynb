```python
import pandas as pd

data = {
    "emp_id": [101, 102, 103, 104, 105],
    "name": ["Asha", "Ravi", "Kiran", "Neha", "Arjun"],
    "department": ["IT", "HR", "IT", "Finance", "HR"],
    "salary": [60000, 45000, 70000, 50000, 48000],
    "experience": [3, 2, 5, 4, 1]
}
df = pd.DataFrame(data)
```

##1. .loc vs .iloc

Task 1.1

Using .loc, select employees who belong to the IT department and display only name and salary.


```python
df.loc[df["department"] == "IT", ["name", "salary"]]
```

Task 1.2

Using .iloc, fetch the first 3 rows and first 2 columns.


```python
df.iloc[:3, :2]
```

Task 1.3

Explain why .iloc[0] works but .loc[0] may fail in some datasets.


`.iloc` uses pure integer-position indexing ($0$ to $N-1$), so `.iloc[0]` always retrieves the first row in memory regardless of the DataFrame's index. `.loc` is strictly label-based; if the DataFrame has custom labels (e.g., employee IDs or strings), or if row 0 was removed after filtering/dropping, label 0 will no longer exist in the index and will raise a KeyError.

##4. Boolean Indexing

Task 4.1

- AND salary > 46000


```python
df[(df["DEPARTMENT"] == "HR") & (df["SALARY"] > 46000)]
```

Task 4.2

```python
df[(df["DEPARTMENT"] == "IT") | (df["EXPERIENCE"] >= 5)]
```
