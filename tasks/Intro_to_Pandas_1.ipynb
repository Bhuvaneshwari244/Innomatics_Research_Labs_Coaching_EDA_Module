##Topic 1: Introduction to Pandas

1. What is Pandas? Explain its features and applications in Data Science.

Definition: Pandas is an open-source Python library built on top of NumPy, designed for flexible, high-performance data manipulation, analysis, and cleaning.

Key Features:
- Primary data structures: 1D Series and 2D tabular DataFrame.  
- Handling missing values (NaN/None) seamlessly.  
- Flexible reshaping, pivoting, slicing, indexing, and subsetting.  
- Robust I/O tools to read and write formats such as CSV, Excel, SQL, and JSON.
- Native group-by and aggregation engines for data munging.

Applications: Exploratory Data Analysis (EDA), feature engineering, time-series analysis, cleaning messy real-world datasets, and preparing data pipelines for machine learning models.

2. Import the Pandas library and explain the purpose of the alias pd.

```python
import pandas as pd
```

Purpose of pd: pd is the globally accepted standard alias/nickname in Python data science. It reduces typing overhead (e.g., pd.DataFrame() instead of pandas.DataFrame()) and maintains code consistency across teams and open-source libraries.

25. Load an Excel workbook containing multiple sheets and read two different sheets using the sheet_name parameter.

```python
%%writefile sheet1.csv
Name,Age,Gender,City,Marks
Asha,21,Female,Hyderabad,85
Rahul,22,Male,Bangalore,92
Kiran,20,Male,Chennai,78
```

```python
%%writefile sheet2.csv
Name,Age,Gender,City,Marks
Neha,23,Female,Mumbai,90
Arjun,21,Male,Delhi,88
Priya,22,Female,Pune,95
```

```python
df1 = pd.read_csv("sheet1.csv")
df2 = pd.read_csv("sheet2.csv")

with pd.ExcelWriter("workbook.xlsx") as writer:
    df1.to_excel(writer, sheet_name="Sheet1", index=False)
    df2.to_excel(writer, sheet_name="Sheet2", index=False)

df_sheet1 = pd.read_excel("workbook.xlsx", sheet_name="Sheet1")
df_sheet2 = pd.read_excel("workbook.xlsx", sheet_name="Sheet2")

print("--- Sheet 1 ---")
display(df_sheet1.head())

print("\n--- Sheet 2 ---")
display(df_sheet2.head())
```
