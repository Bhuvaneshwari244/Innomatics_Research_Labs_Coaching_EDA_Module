## Q1. Normal Indexing on a 1D Array

Input

a = np.array([10,20,30,40,50,60])

2) Print the third element.
3) Print the last element using negative indexing.
4) Print the second-last element.

Expected Output

````python
import numpy as np

# Input array
a = np.array([10, 20, 30, 40, 50, 60])

# Tasks
print(a[0])   # First element -> 10
print(a[2])   # Third element -> 30
print(a[-1])  # Last element using negative indexing -> 60
print(a[-2])  # Second-last element -> 50

# UNDERSTANDING:
# 1. NumPy arrays are 0-indexed, meaning the first element is at index 0.
# 2. Positive indices traverse left-to-right (0, 1, 2, ...).
# 3. Negative indices traverse right-to-left, starting at -1 for the last element.
````

## Q2. Normal Slicing on a 1D Array

Input

a = np.array([10,20,30,40,50,60,70,80])

Tasks

Print

- First four elements
- Last three elements
- Elements from index 2 to index 5
- Every second element
- Reverse the array

Write your understanding of

 start : stop : step


````python
import numpy as np

# Input array
a = np.array([10, 20, 30, 40, 50, 60, 70, 80])

# Tasks
print(a[:4])    # First four elements -> [10 20 30 40]
print(a[-3:])   # Last three elements -> [60 70 80]
print(a[2:6])   # Elements from index 2 to index 5 -> [30 40 50 60]
print(a[::2])   # Every second element -> [10 30 50 70]
print(a[::-1])  # Reverse the array -> [80 70 60 50 40 30 20 10]

# UNDERSTANDING OF start : stop : step:
# 1. 'start': Index where the slice begins (inclusive). Defaults to 0.
# 2. 'stop': Index where the slice ends (exclusive).
# 3. 'step': Step size/interval between items. Negative step reverses direction.
````

````python
import numpy as np

# Input array
a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Tasks
print(a[0])       # First row -> [10 20 30]
print(a[:, 2])    # Third column -> [30 60 90]
print(a[1, 2])    # Element 60 -> 60
print(a[-1])      # Last row -> [70 80 90]
print(a[:2])      # First two rows -> [[10 20 30], [40 50 60]]
print(a[:, -2:])  # Last two columns -> [[20 30], [50 60], [80 90]]
print(a[1:, 1:])  # Sub-matrix 50 60 / 80 90 -> [[50 60], [80 90]]

# UNDERSTANDING:
# 1. 2D indexing syntax is: array[row_index_or_slice, column_index_or_slice].
# 2. Omitting column specification (e.g., a[0]) selects the full row.
# 3. A colon ':' on its own selects all items along that given axis.
````

## Q5. Fancy Indexing on a 1D Array

Input

a = np.array([10,20,30,40,50,60,70])

Tasks

Using Fancy Indexing only, print

- [10 30 50]
- [20 20 40]
- [70 50 30]

Explain what Fancy Indexing is.


````python
import numpy as np

# Input array
a = np.array([10, 20, 30, 40, 50, 60, 70])

# Tasks
print(a[[0, 2, 4]])    # [10 30 50]
print(a[[1, 1, 3]])    # [20 20 40]
print(a[[-1, 4, 2]])   # [70 50 30]

# UNDERSTANDING:
# 1. Fancy indexing involves passing an integer array/list of indices to select arbitrary elements.
# 2. Unlike standard slicing (which creates views), fancy indexing always creates a new independent copy.
````


Tasks

````python
import numpy as np

# Input array
a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Tasks using pure Fancy Indexing:
print(a[[0, 2]])                    # First and third rows -> [[10 20 30], [70 80 90]]
print(a[[0, 2], [1, 2]])            # Elements 20 and 90 -> [20 90]
print(a[[1, 1, 1], [0, 1, 2]])      # Elements 40, 50, 60 using pure coordinate pairs -> [40 50 60]

# UNDERSTANDING:
# 1. Pure fancy indexing pairs each index from row array with the corresponding index in column array (e.g., (1,0), (1,1), (1,2)).
# 2. It always returns a new copy rather than a view.
````

## Q7. Boolean Indexing (1D)

Input

a = np.array([10,15,20,25,30,35,40,45,50])

Tasks

Print

- Values greater than 25
- Values less than 30
- Even numbers
- Odd numbers
- Values between 20 and 40

Explain Boolean Masking.


````python
import numpy as np

# Input array
a = np.array([10, 15, 20, 25, 30, 35, 40, 45, 50])

# Tasks
print(a[a > 25])                      # Values greater than 25 -> [30 35 40 45 50]
print(a[a < 30])                      # Values less than 30 -> [10 15 20 25]
print(a[a % 2 == 0])                  # Even numbers -> [10 20 30 40 50]
print(a[a % 2 != 0])                  # Odd numbers -> [15 25 35 45]
print(a[(a >= 20) & (a <= 40)])       # Values between 20 and 40 (inclusive) -> [20 25 30 35 40]

# UNDERSTANDING:
# 1. Relational operations on arrays create boolean masks containing True and False.
# 2. Combining multiple conditions requires bitwise operators (&, |) with parentheses.
````

## Q9. View Using Slicing

Input

a = np.array([10,20,30,40,50])

````python
import numpy as np

# Input array
a = np.array([10, 20, 30, 40, 50])
b = a[2:]

# Tasks
print("id(a):", id(a))
print("id(b):", id(b))
print("a.ctypes.data:", a.ctypes.data)
print("b.ctypes.data:", b.ctypes.data)
print("a.base:", a.base)
print("b.base:", b.base)

# Modification
b[0] = 999
print("a after modification:", a)  # [ 10  20 999  40  50]
print("b after modification:", b)  # [999  40  50]

# OBSERVATIONS:
# 1. 'a' and 'b' have different id() because 'b' is a separate ndarray wrapper object.
# 2. 'b.base' points to 'a', showing that 'b' is a View sharing memory buffer with 'a'.
# 3. Modifying 'b[0]' directly alters 'a[2]'.
````

## Q10. View Using .view()

Input

a = np.array([10,20,30,40,50])

b = a.view()

Tasks

Repeat the same observations as Question 9.

Then explain

- Is .view() different from slicing?
- Why?


````python
import numpy as np

# Input array
a = np.array([10, 20, 30, 40, 50])
b = a.view()

# Tasks
print("id(a):", id(a))
print("id(b):", id(b))
print("a.ctypes.data:", a.ctypes.data)
print("b.ctypes.data:", b.ctypes.data)
print("a.base:", a.base)
print("b.base:", b.base)

# Modification
b[0] = 888
print("a after modification:", a)  # [888  20  30  40  50]
print("b after modification:", b)  # [888  20  30  40  50]

# UNDERSTANDING:
# 1. Slicing creates a view with an offset memory address (points to the slice start).
# 2. .view() creates a view starting at the exact same memory address (same ctypes.data) and can be used to re-interpret data types.
````

## Q12. View vs Copy Investigation

Create

a = np.array([1,2,3,4,5])

b = a[1:]

c = a.view()

d = a.copy()

Print

id()

ctypes.data

base

Modify all three variables separately.

Observe

- Which changes affect a?
- Which do not?

Write your observations.


````python
import numpy as np

# Inputs
a = np.array([1, 2, 3, 4, 5])
b = a[1:]    # Sliced View
c = a.view() # Explicit View
d = a.copy() # Deep Copy

# Inspection
print("ids:", id(a), id(b), id(c), id(d))
print("ctypes:", a.ctypes.data, b.ctypes.data, c.ctypes.data, d.ctypes.data)
print("bases:", a.base, b.base is not None, c.base is not None, d.base)

# Modifications
b[0] = 200  # Modifies a[1]
c[0] = 100  # Modifies a[0]
d[0] = 500  # Does NOT modify a

print("Final a:", a)  # [100 200   3   4   5]
print("Final d:", d)  # [500   2   3   4   5]

# OBSERVATIONS:
# 1. Modifications to 'b' and 'c' alter 'a' because they share the underlying memory buffer.
# 2. Modifications to 'd' do not affect 'a' because .copy() uses a separate memory buffer.
````

**Normal Indexing**

* **Purpose:** Access continuous elements or slices.


* **Returns:** Single element or `ndarray` (slice).


* **Shares Memory:** Yes (for slicing) / No (for a single scalar element).


* **Creates New ndarray:** Yes (for slicing).


* **Changes Affect Original:** Yes (for slicing).


* **Typical Use Case:** Extracting sub-matrices and row/column ranges.



**Fancy Indexing**

* **Purpose:** Access non-contiguous elements via index arrays or lists.


* **Returns:** Multiple elements (`ndarray`).


* **Shares Memory:** No.


* **Creates New ndarray:** Yes.


* **Changes Affect Original:** No.


* **Typical Use Case:** Extracting specific arbitrary coordinate pairs and batch lookups.



**Boolean Indexing**

* **Purpose:** Filter elements using logical condition masks.


* **Returns:** Multiple elements (1D `ndarray`).


* **Shares Memory:** No.


* **Creates New ndarray:** Yes.


* **Changes Affect Original:** No.


* **Typical Use Case:** Filtering outliers, thresholding, and conditional data cleaning.



**View**

* **Purpose:** Access array data with an alternate layout, stride, or offset.


* **Returns:** Multiple elements (`ndarray`).


* **Shares Memory:** Yes.


* **Creates New ndarray:** Yes.


* **Changes Affect Original:** Yes.


* **Typical Use Case:** Memory-efficient subset processing and heavy matrix manipulation.



**Copy**

* **Purpose:** Create an independent duplicate in memory.


* **Returns:** Multiple elements (`ndarray`).


* **Shares Memory:** No.


* **Creates New ndarray:** Yes.


* **Changes Affect Original:** No.


* **Typical Use Case:** Isolating data transformations and feature engineering safely without mutating raw data.
