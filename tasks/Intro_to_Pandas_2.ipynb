##Topic 7: Row & Column Selection

Q26. Select rows using .loc().

```python
import pandas as pd
```

```python
df_q26 = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Marks": [85, 92, 78, 88, 95],
    "Subject": ["Math", "Science", "History", "Math", "Science"]
}, index=["R1", "R2", "R3", "R4", "R5"])

df_q26
```

```python
# Single row selection
row_single = df_q26.loc["R1"]
row_single
```

```python
# Multiple rows using label slice (inclusive of stop label)
row_slice = df_q26.loc["R2":"R4"]
row_slice
```

```python
# Conditional row selection using boolean mask
row_condition = df_q26.loc[df_q26["Marks"] > 85]
row_condition
```

Q27. Select rows using .iloc().

```python
df_q27 = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Marks": [85, 92, 78, 88, 95],
    "Subject": ["Math", "Science", "History", "Math", "Science"]
})
df_q27
```

```python
# Single row by integer position
row_pos = df_q27.iloc[0]
row_pos
```

```python
# Range of rows by integer slice (exclusive of stop index)
rows_range = df_q27.iloc[1:4]
rows_range
```

```python
# Arbitrary rows using a list of positions
rows_list = df_q27.iloc[[0, 2, 4]]
rows_list
```

Q28. Demonstrate the difference between .loc() and .iloc() using the same DataFrame.


```python
df_q28 = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Marks": [85, 92, 78, 88]
}, index=[10, 20, 30, 40])
df_q28
```

```python
# .loc searches for the literal label 30 and INCLUDES it
loc_demo = df_q28.loc[10:30]
loc_demo
```

```python
# .iloc counts 0-based integer positions and EXCLUDES stop index 3
iloc_demo = df_q28.iloc[0:3]
iloc_demo
```
