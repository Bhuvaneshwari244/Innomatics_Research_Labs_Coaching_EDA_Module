````python
import pandas as pd
import numpy as np

# Base dataset
df = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Rahul", "Priya", "Arjun", "Sneha", "Kiran", "Pooja", "Vikram", "Anjali"],
    "Department": ["DS", "HR", "IT", "Finance", "IT", "HR", "DS", "Finance"],
    "City": ["Hyderabad", "Bangalore", "Chennai", "Mumbai", "Hyderabad", "Bangalore", "Chennai", "Mumbai"],
    "Salary": [50000, 45000, 60000, 55000, 70000, 48000, 75000, 65000],
    "Experience": [2, 4, 5, 3, 8, 4, 9, 6],
    "Rating": [4, 3, 5, 4, 5, 3, 5, 4]
})

display(df)
````

##🔹 Section A: Filtering


###1. Basic Filtering

###2. Multiple Condition Filtering

1. Display employees from the IT department whose salary is greater than 60000.


````python
display(df[(df["Department"] == "IT") & (df["Salary"] > 60000)])
````

4. Display employees from Hyderabad or Chennai.


````python
display(df[(df["City"] == "Hyderabad") | (df["City"] == "Chennai")])
````

##🔹 Section E: Apply Function


###9. Apply Function Practice


````python
def add_ten_percent(val):
    return val * 1.10

def calc_annual(val):
    return val * 12

def calc_bonus(val):
    return val * 0.10

def round_increased_salary(val):
    return round(val * 1.10)

def increment_exp(val):
    return val + 1

def square_val(val):
    return val ** 2
````

1. Increase salary by 10%.


````python
display(df["Salary"].apply(add_ten_percent))
````

2. Calculate Annual Salary.


````python
display(df["Salary"].apply(calc_annual))
````

3. Calculate Bonus equal to 10% of Salary.

````python
display(df["Salary"].apply(calc_bonus))
````

4. Round salary values after increasing them by 10%.


````python
display(df["Salary"].apply(round_increased_salary))
````

5. Add 1 year to every employee's Experience.


````python
display(df["Experience"].apply(increment_exp))
````

6. Find the square of Experience values.


````python
display(df["Experience"].apply(square_val))
````

##🔹 Section F: Lambda Functions


###10. Lambda Practice


````python
df_lambda = df.copy()
display(df_lambda)
````

1. Increase salary by 10% using lambda.


````python
df_lambda["Increased_Salary"] = df_lambda["Salary"].apply(lambda x: x * 1.10)
display(df_lambda)
````

2. Create an Annual_Salary column using lambda.


````python
df_lambda["Annual_Salary"] = df_lambda["Salary"].apply(lambda x: x * 12)
display(df_lambda)
````

3. Create a Bonus column equal to 10% of Salary.


````python
df_lambda["Bonus"] = df_lambda["Salary"].apply(lambda x: x * 0.10)
display(df_lambda)
````

4. Create a Tax column equal to 5% of Salary.


````python
df_lambda["Tax"] = df_lambda["Salary"].apply(lambda x: x * 0.05)
display(df_lambda)
````

5. Add 1 year to every employee's Experience using lambda.


````python
df_lambda["New_Experience"] = df_lambda["Experience"].apply(lambda x: x + 1)
display(df_lambda)
````

##🔹 Section G: Map Function


###11. Mapping Practice


````python
df_map = df.copy()

dept_mapping = {
    "DS": "D01",
    "HR": "D02",
    "IT": "D03",
    "Finance": "D04"
}

city_mapping = {
    "Hyderabad": "HYD",
    "Bangalore": "BLR",
    "Chennai": "MAA",
    "Mumbai": "BOM"
}
````

````python
df_map["Department_Code"] = df_map["Department"].map(dept_mapping)
df_map
````

2. Create a City_Code column using map().


````python
df_map["City_Code"] = df_map["City"].map(city_mapping)
df_map
````
