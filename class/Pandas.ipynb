Filtering: Dataframe

```python
arr = np.arange(1, 21).reshape(4, 5)
arr
```

```python
df = pd.DataFrame(arr)
df
```

```python
df = pd.DataFrame(arr, index=['r1', 'r2', 'r3', 'r4'])
df
```

```python
df = pd.DataFrame(arr, index=['r1', 'r2', 'r3', 'r4'], columns=['c1', 'c2', 'c3', 'c4', 'c5'])
df
```

By applying a condition on an column
- Get Booleans on rows
- Filter the rows of a datafram

```python
# Filter column c1 with `>6`
#df['c1'] or df.c1
df['c1'] > 6
```

```python
df.loc[df['c1'] > 6]
```

```python
# On col 'c2' condition - %2 == 0
df['c2']%2 == 0
```

```python
df[df['c2']%2 == 0]
```

```python
# Filter the elements of column 'c4' either divisible by 2 or divisible by 3
(df.c4%2 == 0) | (df.c4%3 == 0)
```

```python
df[(df.c4%2 == 0) | (df.c4%3 == 0)]
```

```python
# Get the Negation of aove result (use tilde)
df[~((df.c4%2 == 0) | (df.c4%3 == 0))]
```

```python
~((df.c4%2 == 0) | (df.c4%3 == 0))
```

```python
df
```

```python
~((df.c5%5 == 0) & (df.c5%10 == 0))
```

```python
df[~((df.c5%5 == 0) & (df.c5%10 == 0))]
```

```python
# get only columns c2 and c4
# Traditional way - NOT RECOMMENDED - chaining
df[~((df.c5%5 == 0) & (df.c5%10 == 0))][['c2', 'c4']]
```

df.loc[rI/RRange/Rlist/**RBool**, cI/Crange/**clist**/cBool]
- Apply a condition on columns
  - returns Row Bool

```python
df.loc[~((df.c5%5 == 0) & (df.c5%10 == 0)), ['c2', 'c4']]
```

```python
df
```
