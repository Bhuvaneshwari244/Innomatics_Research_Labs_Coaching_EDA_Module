```python
# Replace [9, 10] with [90, 100]
arr1[2, 0:2] = [90, 100]
arr1
```

```python
# replace [7, 11] with [70, 110]
arr1[1:3, 2] = [70, 110]
arr1
```

```python
# Replace [4, 8] with [40, 80]
arr1[0:2, 3:4]
```

```python
arr1[0:2, 3:4] = [[40], [80]]
arr1
```

```python
# Replace [2, 5] with [20, 50]
arr1[[0, 1], [1, 0]] = [20, 50]
arr1
```

```python
# Replace second column with [2, 6, 10]
arr1[:, 1]
```

```python
# Replace second column with [2, 6, 10]
arr1[:, 1] = [2, 6, 10]
arr1
```

```python
# Replace the 3rd row with all 50s
arr1[2]
```

```python
# Replace the 3rd row with all 50s
arr1[2] = 50 # Broadcasting (1, 4) with (1,)
arr1
```

```python
# Replace second column with all 100
arr1[:, 1]
```

```python
arr1[:, 1:2] = 100 # Broadcasting (3, 1) with (1,)
```

Inserting elements for - 1D, rows/columns for 2D

`np.insert(arr, index/list of index, ele/list or array of elelemnts)`

```python
# 1D
arr1 = np.arange(1, 11, 2)
arr1
```

```python
# Insert 0 at index 0
np.insert(arr1, 0, 0) # creates a copy
```

```python
arr1
```

```python
# Insert 10 at last index
arr1 = np.insert(arr1, -1, 10)
arr1
```

```python
# Insert 10 at last index
arr1 = np.insert(arr1, 7, 10)
arr1
```

```python
# Insert 2 and 4 at locations [2, 4]
arr1 = np.insert(arr1, [2, 4], [2, 4])
arr1
```

uses relative positions to insert elements

```python
arr2 = np.arange(1, 5).reshape(2, 2)
arr2
```

```python
# Add a new row [5, 6]
np.insert(arr2, 2, [5, 6])
```

`axis`

```python
# Add a new row [5, 6]
np.insert(arr2, 2, [5, 6], axis = 0)
```

```python
# Add a new row [5, 6]
np.insert(arr2, 2, [5, 6], axis = 1)
```

```python
arr2 = np.arange(1, 13).reshape(3, 4)
arr2
```

```python
# Insert a row at index 2 with values all 50
np.insert(arr2, 2, 50, axis = 0)
```

```python
# Insert a column at index 1 +-elements as 100
np.insert(arr2, 1, 100, axis = 1)
```

```python
# Insert a row at index 2 with elements [100, 200, 300, 400]
np.insert(arr2, 2, [100, 200, 300, 400], axis = 0)
```

```python
# Insert a column at index 2 with elements [100, 200, 300]
np.insert(arr2, 2, [100, 200, 300], axis = 1)
```
