for loop

```python
l1 = [1, 2, 3]
l2 = [4, 5, 6]

out = []
for e1, e2 in zip(l1, l2):
  out.append(e1+e2)
out
```

comprehension

```python
out = [e1+e2 for e1, e2 in zip(l1, l2)]
out
```
