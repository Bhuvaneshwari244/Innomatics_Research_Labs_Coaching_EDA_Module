## Q1. Integer Data Types and Memory

Create the array [10, 20, 30, 40, 50] using the following dtypes:
- np.int8
- np.int16
- np.int32
- np.int64

For each array:
- Print dtype
- Print itemsize
- Print nbytes

Based on your observations, explain how memory usage changes with different integer data types and when you would prefer each datatype.


```python
import numpy as np

data = [10, 20, 30, 40, 50]
int_dtypes = [np.int8, np.int16, np.int32, np.int64]

for dt in int_dtypes:
    arr = np.array(data, dtype=dt)
    print(f"dtype: {arr.dtype} | itemsize: {arr.itemsize} bytes | nbytes: {arr.nbytes} bytes")
```

Memory Usage & Preferences:
- Memory Scaling: Total memory increases directly with bit-width: nbytes = number of elements × itemsize.
- np.int8 (-128 to 127): Best for bounded, small integer ranges (e.g., human age, RGB channels, image masks) where memory conservation is critical.
- np.int16 (-32,768 to 32,767): Best for medium-range counts, sensor telemetry, and 16-bit audio samples.
- np.int32 (-2.14B to 2.14B): Standard default for typical data science counting, indexing, and tabular data.
- np.int64 : Essential for large IDs, timestamps (epoch milliseconds/nanoseconds), and big-data computations requiring overflow protection.

## Q2. Floating Point Data Types and Object Arrays

Create the array [10.5, 20.5, 30.5] using:
- np.float16
- np.float32
- np.float64

Also create the following array using dtype=object:

["Python", 100, 3.14, True]

Print the dtype, itemsize, and nbytes of each array.

Based on your observations, explain:
- Which float datatype provides the highest precision?
- Which float datatype consumes the least memory?
- Why is dtype=object required for mixed data?


```python
float_dtypes = [np.float16, np.float32, np.float64]
data_f = [10.5, 20.5, 30.5]

for dt in float_dtypes:
    arr = np.array(data_f, dtype=dt)
    print(f"dtype: {arr.dtype} | itemsize: {arr.itemsize} bytes | nbytes: {arr.nbytes} bytes")

obj_arr = np.array(["Python", 100, 3.14, True], dtype=object)
print(f"dtype: {obj_arr.dtype} | itemsize: {obj_arr.itemsize} bytes | nbytes: {obj_arr.nbytes} bytes")
```

Key Findings:

- Highest Precision: np.float64 (double precision, ~15–17 decimal digits).

- Least Memory: np.float16 (half precision, 2 bytes per item).

- Why dtype=object for Mixed Data: NumPy arrays require homogeneous, contiguous memory. When element types differ, NumPy uses dtype=object to store 64-bit memory pointers referencing standard Python heap objects rather than raw contiguous primitives.

# Q4. Array Properties

Create any 3D NumPy array.

Print and explain the purpose of:

- dtype
- shape
- ndim
- size
- itemsize
- nbytes

Also explain how these properties help while working with large datasets.


```python
arr_3d = np.ones((2, 3, 4), dtype=np.int32)

print("dtype:   ", arr_3d.dtype)
print("shape:   ", arr_3d.shape)
print("ndim:    ", arr_3d.ndim)
print("size:    ", arr_3d.size)
print("itemsize:", arr_3d.itemsize)
print("nbytes:  ", arr_3d.nbytes)
```

Property Functions & Utility in Large Datasets:
- dtype : Identifies the element data format; prevents silent type promotion and numeric overflow.

- ndim : Reports total axes count; confirms model input rank requirements.
- size : Total element count (∏ shape); used to compute iteration boundaries and flatten operations.
- itemsize : Byte size of a single value; helps profile per-unit memory footprint.
- nbytes : Total memory consumed (size × itemsize); critical for preventing out-of-memory (OOM) errors during large batch loads.

# Q5. NumPy Random Functions

Generate random values using the following functions:
- np.random.random()
- np.random.rand()
- np.random.randn()
- np.random.randint()
- np.random.normal()

Compare the outputs and explain:
- Range of values
- Distribution
- Difference between each function
- One real-world use case for each function


```python
np.random.seed(42)

print("random():", np.random.random(size=(2, 2)))
print("rand():  ", np.random.rand(2, 2))
print("randn(): ", np.random.randn(2, 2))
print("randint():", np.random.randint(1, 10, size=(2, 2)))
print("normal():", np.random.normal(loc=0.0, scale=1.0, size=(2, 2)))
```

**`np.random.random()`**

* **Range:** $[0.0, 1.0)$

* **Distribution:** Continuous Uniform distribution


* **Parameter Syntax:** Accepts dimensions as a single shape tuple: `size=(d1, d2, ...)`

* **Real-World Use Case:** Generating random dropout masks or probabilistic thresholds in machine learning pipelines.


**`np.random.rand()`**

* **Range:** $[0.0, 1.0)$

* **Distribution:** Continuous Uniform distribution


* **Parameter Syntax:** Accepts dimensions as separate positional arguments: `d0, d1, ...`

* **Real-World Use Case:** Quickly generating synthetic feature matrices or normalized pixel test data.


**`np.random.randn()`**

* **Range:** $(-\infty, \infty)$

* **Distribution:** Standard Normal (Gaussian) distribution ($\mu=0, \sigma=1$)


* **Parameter Syntax:** Accepts dimensions as separate positional arguments: `d0, d1, ...`

* **Real-World Use Case:** Random weight initialization for neural network layers to avoid vanishing/exploding gradients.


**`np.random.randint()`**

* **Range:** Discrete values from $\text{low}$ (inclusive) to $\text{high}$ (exclusive)


* **Distribution:** Discrete Uniform distribution


* **Parameter Syntax:** `low, high, size=(...)`

* **Real-World Use Case:** Bootstrapping, random sampling of dataset row indices, or rolling virtual dice.


**`np.random.normal()`**

* **Range:** $(-\infty, \infty)$

* **Distribution:** Gaussian (Normal) distribution with configurable mean (`loc`) and standard deviation (`scale`)


* **Parameter Syntax:** `loc, scale, size=(...)`

* **Real-World Use Case:** Simulating real-world measurement errors, stock price shocks, or sensor noise.

# Q6. Shuffle vs Permutation

Create an array containing numbers from 1 to 10.

Apply:
- np.random.shuffle()
- np.random.permutation()

Compare the outputs and explain:
- Which function modifies the original array?
- Which returns a new array?
- When would you use each one?


```python
arr1 = np.arange(1, 11)
arr2 = np.arange(1, 11)

res_shuffle = np.random.shuffle(arr1)
res_perm = np.random.permutation(arr2)

print("Original after shuffle:    ", arr1)
print("Return of shuffle:         ", res_shuffle)
print("Original after perm:       ", arr2)
print("Return of permutation:     ", res_perm)
```

Comparison:

np.random.shuffle(): Modifies the array in-place and returns None. Best used when managing large in-memory datasets where duplicating data causes unnecessary memory overhead.

np.random.permutation(): Leaves the input untouched and returns a new, shuffled copy. Best used when preserving the original baseline dataset for cross-validation or subsequent pipelines is required.

## Q10. Python List vs NumPy Array Comparison

Create the following using both a Python List and a NumPy Array.

[10, 20, 30, 40, 50]

Perform the following operations on both and compare the results.

Tasks
1) Print the type of both objects.

2) Print the length of both objects.

3) Add 10 to every element and observe the output.

4) Multiply every element by 2 and compare the results.

5) Access the third element from both.

6) Slice elements from index 1 to 4.

7) Check the memory occupied by both objects.

8) Perform element-wise addition of two lists/arrays.

9) Based on your observations, explain why NumPy Arrays are preferred over Python Lists in Data Science and Machine Learning.


```python
import sys

py_list = [10, 20, 30, 40, 50]
np_arr = np.array([10, 20, 30, 40, 50])

# 1. Type
print("Type:", type(py_list), "vs", type(np_arr))

# 2. Length
print("Length:", len(py_list), "vs", len(np_arr))

# 3. Add 10
# py_list + 10 raises TypeError; requires list comprehension
print("Add 10:", [x + 10 for x in py_list], "vs", np_arr + 10)

# 4. Multiply by 2
print("Multiply 2:", py_list * 2, "vs", np_arr * 2)

# 5. Access 3rd element
print("3rd Element:", py_list[2], "vs", np_arr[2])

# 6. Slicing index 1 to 4
print("Slice [1:4]:", py_list[1:4], "vs", np_arr[1:4])

# 7. Memory
py_list_mem = sys.getsizeof(py_list) + sum(sys.getsizeof(x) for x in py_list)
print(f"Memory: Python List ~{py_list_mem} bytes vs NumPy Array {np_arr.nbytes} bytes")

# 8. Element-wise addition
py_list2 = [1, 2, 3, 4, 5]
np_arr2 = np.array([1, 2, 3, 4, 5])
print("Vector Add:", [x + y for x, y in zip(py_list, py_list2)], "vs", np_arr + np_arr2)
```

Why NumPy Outperforms Python Lists in DS/ML:

- Vectorized Computations: NumPy eliminates Python for loops by dispatching operations directly to pre-compiled C/Fortran SIMD vector instructions.

- Contiguous Memory Storage: NumPy stores primitive bytes in continuous memory blocks, whereas Python lists store scattered pointer references to full-overhead PyObject wrappers.

- Broadcasting & Matrix Operations: Provides native multidimensional broadcasting and tensor math essential for linear algebra and deep learning.
