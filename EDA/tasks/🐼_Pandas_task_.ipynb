#🐼 Pandas Series – Practice Questions

1. Create a Pandas Series from a list of daily temperatures.

```python
temps = [32.5, 34.0, 31.8, 30.2, 33.1]
s_temps = pd.Series(temps)
print(s_temps)
```

2. Create a Pandas Series from a tuple of employee names.


```python
employees = ("Alice", "Bob", "Charlie", "David")
s_emp = pd.Series(employees)
print(s_emp)
```

3. Create a Pandas Series from a set of unique product IDs.


```python
product_ids = {101, 102, 103, 104}
s_products = pd.Series(list(product_ids))
print(s_products)
```

4. Create a Pandas Series from a dictionary where keys are subjects and values are marks.


```python
marks_dict = {"Math": 95, "Physics": 88, "Chemistry": 92, "English": 85}
s_marks = pd.Series(marks_dict)
print(s_marks)
```

5. Create a Pandas Series from a NumPy array of monthly expenses.

```python
import numpy as np
expenses = np.array([1200, 1500, 1100, 1800, 1400])
s_expenses = pd.Series(expenses)
print(s_expenses)
```

6. Print the index and values of the Series created from the list.


```python
print("Index:", s_temps.index)
print("Values:", s_temps.values)
```

7. Print the index and values of the Series created from the tuple.


```python
print("Index:", s_emp.index)
print("Values:", s_emp.values)
```

8. Print the index and values of the Series created from the dictionary.


```python
print("Index:", s_marks.index)
print("Values:", s_marks.values)
```

9. Create a Series using a list and manually assign custom index labels.


```python
data = [10, 20, 30, 40]
custom_index = ['a', 'b', 'c', 'd']
s_custom = pd.Series(data, index=custom_index)
print(s_custom)
```

10. Create a Series from a dictionary and display only its values.

```python
sample_dict = {'item1': 50, 'item2': 75, 'item3': 100}
s_dict = pd.Series(sample_dict)
print(s_dict.values)
```

11. Convert a list of Boolean values into a Pandas Series.


```python
bool_list = [True, False, True, True, False]
s_bool = pd.Series(bool_list)
print(s_bool)
```

12. Create a Series from a NumPy array and check its data type.


```python
np_arr = np.array([1.5, 2.7, 3.9, 4.1])
s_arr = pd.Series(np_arr)
print("Series dtype:", s_arr.dtype)
```

# 🐼 Pandas DataFrame – Practice Questions


13. Create a Pandas DataFrame from a list of lists representing student ID, name, and score.


```python
students = [
    [101, "John", 85],
    [102, "Emma", 92],
    [103, "Alex", 78]
]
df_students = pd.DataFrame(students)
print(df_students)
```

14. Create a Pandas DataFrame from a dictionary of lists representing product, price, and quantity.


```python
products_data = {
    "Product": ["Laptop", "Mouse", "Keyboard"],
    "Price": [1200, 25, 75],
    "Quantity": [5, 20, 15]
}
df_products = pd.DataFrame(products_data)
print(df_products)
```

15. Create a Pandas DataFrame from a NumPy array representing employee data.


```python
emp_data = np.array([
    [1, "Tech", 70000],
    [2, "HR", 55000],
    [3, "Finance", 65000]
])
df_emp = pd.DataFrame(emp_data)
print(df_emp)
```

16. Assign appropriate column names to the DataFrame created from the list of lists.


```python
df_students.columns = ["StudentID", "Name", "Score"]
print(df_students)
```

17. Assign column names to the DataFrame created from the NumPy array.


```python
df_emp.columns = ["EmpID", "Department", "Salary"]
print(df_emp)
```

18. Print the shape of each DataFrame created.

```python
print("Students Shape:", df_students.shape)
print("Products Shape:", df_products.shape)
print("Employees Shape:", df_emp.shape)
```

19. Print the column names of each DataFrame.


```python
print("Students Columns:", df_students.columns.tolist())
print("Products Columns:", df_products.columns.tolist())
print("Employees Columns:", df_emp.columns.tolist())
```

20. Print the index values of one DataFrame.


```python
print("Products Index:", df_products.index)
```

21. Display the complete structure (rows and columns) of a DataFrame using attributes.


```python
print("Index (Rows):", df_products.axes[0])
print("Columns:", df_products.axes[1])
print("Dimensions (Rows, Cols):", df_products.shape)
```
