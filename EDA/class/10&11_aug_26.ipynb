### Copy vs view:

why views are required?
- In the data analytics, dealing with very large datasets
- Dataset of size 2 GB -> like 10 Lakh rows & 1 Lakh column
- perform analysis on datset -> apply operations on analysis
- Dataset ->(moved)-> RAM

- Perform operations -> some ele in dataset -> changed -> keeping original copy as it -> To create a new copy further 2GB request -> But only 0.5GB available
- Cannot create a new copy -> To apply transformation -> Apply transformation 'inplace' -> on the views of any array
1) Memory saved
2) Quick computation

Computation of list vs ndarray
- take a list/ndarray with 10,00,000 elements in an order, multiply 2 with every element

```python
import time
```

```python
lst = list(range(0, 10_00_000))
lst
```

```python
lst * 2 # repetition
```

```python
# multiplication
lst_out = []
for ele in lst:
    lst_out.append(ele * 2)
lst_out
```

```python
lst_out = []
start = time.time()
for ele in lst:
    lst_out.append(ele * 2)
end = time.time()
elapsed_time = end-start
elapsed_time
```

```python
arr = np.arrange(0, 10_00_00)
```

```python
start = time.time()

arr_out = arr * 2
end = time.time()
elapsed_time = end-start
elapsed_time # sec
```
