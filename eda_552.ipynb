**Context Manager**
- NO NEED TO CLOSE THE FILE
- Used with `with`
- Implemented using a protocol - `context manager protocol`
- In a class implementing dunder methods
  - `__enter__`
  - `__exit__`

- Used to Create an object of any resouce
- Carry out operations on resource and
- Close the resource automatically

`````python
with open('file4.txt', 'w') as f: # Creating an Object
  f.writelines(['Hello\n', 'World\n'])
`````

`````python
f.closed
`````

`````python
# Implemetation COntext manager Protocol
class MyClass:
  def __init__(self):
    self.obj = None
  def __enter__(self):
    print('Entering into context')
    return self.obj
  def __exit__(self, exec_name = None, exec_value = None, exec_traceback = None):
    print('Exiting from context')
    if exec_name or exec_value or exec_traceback:
      print(f'Name: {self.exec_name}, Value: {self.exec_value} and Traceback: {self.exec_traceback}')
    return True  # Silenced by Python
    # return False # To manually handle error
`````

`````python
with MyClass() as obj:
  print('Inside the context manager')
`````

`````python
import pandas as pd
import seaborn as sns
import scipy.stats as st
`````

`````python
tips = sns.load_dataset('tips')
tips.head()
`````

**Claim on single Variable**: The Avg total bill paid by customers of a Restaurent is $20

H0: mean = $20

H1: mean != $20

`````python
# One sample t-test
st.ttest_1samp(tips['total_bill'], 20)
`````

Result p(0.70) > 0.05
- Unable to Reject Null Hypothesis(mean is 20)
- Accepting the mean bill paid by customers in any restaurent is $20

**Claim on Categorical Variable**: The proportion of smokers and non smokers is 50-50

H0: P(smoker) = 0.5

H1: P(SMoker) != 0.5

`````python
tips['smoker'].value_counts()
`````

`````python
# Chi square - GOODNESS OF FIT TEST
st.chisquare(tips['smoker'].value_counts())
`````

Result: p(0) < 0.05
- Reject the Null hypothesis
- The Proportion of smokers and Non smokers are not same

**Claim on two Numeric Variables**: Tip and Total bill are correlated

DEFAULT HYPOTHESIS

H0: Variables are not correlated (rho = 0)

H1: Variables are correlated (rho != 0)

`````python
# Assumption - LINEAR RELATIONSHIP BETWEEN VARIABLES
sns.scatterplot(data = tips, x = 'tip', y = 'total_bill')
sns.regplot(data = tips, x = 'tip', y = 'total_bill')
`````

`````python
# Pearsonr test
st.pearsonr(tips['total_bill'], tips['tip'])
`````

`````python
# spearmanr test
st.spearmanr(tips['total_bill'], tips['tip'])
`````

Result: p(0) < 0.05

- Tip and Total bill are correlated

**Claim on two categorical Variables:** Sex and Smoker are independent

H0: Sex and Smoker are Independent

H1: Sex and Smoker are Dependent

`````python
ct = pd.crosstab(tips['sex'], tips['smoker'])
ct
`````

`````python
# Chisquare test for independence
st.chi2_contingency(observed = ct)
`````

Result p(1) > 0.05
- Unable to Reject Null Hypothesis (Independent)
- Sex and Smoker are Independent

**Claim on a numeric variable and a categorical variable**: Total bill paid by Smokers and Non smokers are same

H0: mean_tb_smokers = mean_tb_non_smokers

H1: mean_tb_smokers != mean_tb_non_smokers

`````python
tips.head()
`````

`````python
# Independent Samples t-test
## NORMALITY - ShapiroWilk test
## Homoscedasticity - Levene test
`````

`````python
tb_smokers = tips.loc[tips['smoker'] == 'Yes', 'total_bill']
tb_non_smokers = tips.loc[tips['smoker'] == 'No', 'total_bill']
`````

Normality: DEFAULT HYPOTHESIS

H0: Data is Normally Distributed

H1: Data is Not Normally Distributed

`````python
# Normality
st.shapiro(tb_smokers)
`````

`````python
st.shapiro(tb_non_smokers)
`````

Result p(0) < 0.05
- Reject Null Hypothesis
- Data is not normally distributed

**To make data follow nearly normal - Apply Data Transformations**

Homoscedasticity: DEFAULT Hypothesis

H0: Variaances are equal

H1: Variances are not equal

`````python
st.levene(tb_smokers,tb_non_smokers)
`````

Result p(0.45) < 0.05
- Reject Null Hypothesis
- Data Total bill smokers and Non smokers are not Homoscedastic

**To make data Homoscedastic - Apply Data Transformations**

`````python
# Assume - Both Assumptions are Satisfied
## Apply Independent samples t-test
st.ttest_ind(tb_smokers, tb_non_smokers)
`````

Result: p(0.18) > 0.05
- Unable to Reject Null Hypothesis
- Total bill paid by Smokers and Non smokers are same

**Because assumptions are not satisfied use a non parametric test names as: Mannwhitney U-test**

`````python
st.mannwhitneyu(tb_smokers, tb_non_smokers)
`````

Result: p(0.34) > 0.05
- Unable to Reject Null Hypothesis
- Total bill paid by Smokers and Non smokers are same

**Claim on Multiple variables:** Average total bill paid by the customers during all the week days are same


H0: mean_tb_thrs = mean_tb_fri = mean_tb_sat = mean_tb_sun

H1: mean_tb_thrs != mean_tb_fri != mean_tb_sat != mean_tb_sun

`````python
# Oneway ANOVA
## Normality - Shapiro wilk
## Homoscedasticity - Levene
### ASSUMING BOTH ARE SATISFIED
`````

`````python
tips.day.unique()
`````

`````python
tb_thur = tips.loc[tips['day'] == 'Thur', 'total_bill']
tb_fri = tips.loc[tips['day'] == 'Fri', 'total_bill']
tb_sat = tips.loc[tips['day'] == 'Sat', 'total_bill']
tb_sun = tips.loc[tips['day'] == 'Sun', 'total_bill']
`````

`````python
# Oneway ANOVA
st.f_oneway(tb_thur, tb_fri, tb_sat, tb_sun)
`````

Result: p(0.042) < 0.05
